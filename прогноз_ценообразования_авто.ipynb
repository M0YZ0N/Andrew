```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix, mean_absolute_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# 1. Генерация искусственного датасета автомобилей
rng = np.random.default_rng(42)
n = 2000  # количество автомобилей

# Основные характеристики
horsepower = rng.normal(150, 50, size=n)  # мощность в л.с.
horsepower = np.clip(horsepower, 50, 400)  # ограничиваем диапазон

mileage = rng.normal(80000, 40000, size=n)  # пробег в км
mileage = np.clip(mileage, 0, 300000)

year = rng.integers(2000, 2024, size=n)  # год выпуска

engine_volume = rng.normal(2.0, 0.6, size=n)  # объем двигателя в литрах
engine_volume = np.clip(engine_volume, 1.0, 5.0)

# Категориальные признаки
transmission = rng.choice(['automatic', 'manual'], size=n, p=[0.6, 0.4])
fuel_type = rng.choice(['petrol', 'diesel', 'hybrid', 'electric'],
                       size=n, p=[0.5, 0.3, 0.15, 0.05])
car_type = rng.choice(['sedan', 'suv', 'hatchback', 'coupe'],
                      size=n, p=[0.4, 0.3, 0.2, 0.1])

# Базовая цена в зависимости от характеристик
base_price = (
    500_000 * (engine_volume / 2.0) +  # зависимость от объема двигателя
    1000 * horsepower +                 # зависимость от мощности
    50_000 * ((year - 2000) / 24) -     # зависимость от года
    0.002 * mileage +                   # зависимость от пробега
    200_000 * (transmission == 'automatic') +  # автоматическая коробка дороже
    150_000 * (fuel_type == 'hybrid') +        # гибрид дороже
    100_000 * (fuel_type == 'electric') -      # электромобили пока дешевле
    50_000 * (fuel_type == 'diesel')           # дизель дешевле бензина
)

# Добавляем шум и дополнительные факторы
noise = rng.normal(0, 200_000, size=n)
price = np.clip(base_price + noise, 200_000, 5_000_000)  # ограничиваем диапазон цен

# Создаем DataFrame
df = pd.DataFrame({
    'price': price.round(0),
    'horsepower': horsepower.round(0),
    'mileage': mileage.round(0),
    'year': year,
    'engine_volume': engine_volume.round(1),
    'transmission': transmission,
    'fuel_type': fuel_type,
    'car_type': car_type,
})

# 2. Предобработка данных
# Кодируем категориальные признаки
label_encoders = {}
for column in ['transmission', 'fuel_type', 'car_type']:
    le = LabelEncoder()
    df[column + '_encoded'] = le.fit_transform(df[column])
    label_encoders[column] = le

# 3. Разделение данных на признаки и целевую переменную
# Для классификации: разбиваем цену на 3 класса
q1, q2 = df['price'].quantile([0.33, 0.66])

def price_to_class(p):
    if p <= q1:
        return 0  # бюджетный
    if p <= q2:
        return 1  # средний класс
    return 2      # премиум

df['price_class'] = df['price'].apply(price_to_class)

# Подготовка признаков
features = ['horsepower', 'mileage', 'year', 'engine_volume',
            'transmission_encoded', 'fuel_type_encoded', 'car_type_encoded']

X = df[features].values
y_class = df['price_class'].values
y_reg = df['price'].values  # для регрессии

# Разделение на обучающую и тестовую выборки
X_train, X_test, y_train_class, y_test_class, y_train_reg, y_test_reg = train_test_split(
    X, y_class, y_reg, test_size=0.3, random_state=42, stratify=y_class
)

# Масштабирование признаков
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Нейронная сеть для классификации ценовых категорий
print("\n" + "="*50)
print("МОДЕЛЬ КЛАССИФИКАЦИИ АВТОМОБИЛЕЙ")
print("="*50)

model_class = Sequential([
    Dense(64, activation='relu', input_shape=(X_train_scaled.shape[1],)),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(3, activation='softmax')  # 3 класса
])

model_class.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Ранняя остановка для предотвращения переобучения
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True
)

# Обучение модели
history_class = model_class.fit(
    X_train_scaled, y_train_class,
    validation_split=0.2,
    epochs=50,
    batch_size=32,
    verbose=1,
    callbacks=[early_stopping]
)

# Оценка модели классификации
loss_class, acc_class = model_class.evaluate(X_test_scaled, y_test_class, verbose=0)
print(f'\nТочность классификации: {acc_class:.4f}')

# Прогнозы для классификации
y_pred_class = model_class.predict(X_test_scaled)
y_pred_class_labels = np.argmax(y_pred_class, axis=1)

print("\nОтчет по классификации:")
print(classification_report(y_test_class, y_pred_class_labels,
                           target_names=['Бюджет', 'Средний', 'Премиум']))

# 5. Нейронная сеть для регрессии (предсказание конкретной цены)
print("\n" + "="*50)
print("МОДЕЛЬ РЕГРЕССИИ ДЛЯ ПРОГНОЗА ЦЕНЫ")
print("="*50)

# Масштабирование целевой переменной для регрессии
scaler_price = StandardScaler()
y_train_reg_scaled = scaler_price.fit_transform(y_train_reg.reshape(-1, 1)).flatten()
y_test_reg_scaled = scaler_price.transform(y_test_reg.reshape(-1, 1)).flatten()

model_reg = Sequential([
    Dense(128, activation='relu', input_shape=(X_train_scaled.shape[1],)),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(16, activation='relu'),
    Dense(1, activation='linear')  # один выход для регрессии
])

model_reg.compile(
    optimizer='adam',
    loss='mse',  # среднеквадратичная ошибка для регрессии
    metrics=['mae']  # средняя абсолютная ошибка
)

# Обучение модели регрессии
history_reg = model_reg.fit(
    X_train_scaled, y_train_reg_scaled,
    validation_split=0.2,
    epochs=100,
    batch_size=32,
    verbose=1,
    callbacks=[early_stopping]
)

# Оценка модели регрессии
loss_reg, mae_reg = model_reg.evaluate(X_test_scaled, y_test_reg_scaled, verbose=0)

# Обратное масштабирование предсказаний
y_pred_reg_scaled = model_reg.predict(X_test_scaled)
y_pred_reg = scaler_price.inverse_transform(y_pred_reg_scaled).flatten()

# Метрики регрессии
mae = mean_absolute_error(y_test_reg, y_pred_reg)
r2 = r2_score(y_test_reg, y_pred_reg)

print(f'\nMAE (Средняя абсолютная ошибка): {mae:.2f} руб.')
print(f'R² (Коэффициент детерминации): {r2:.4f}')

# 6. Визуализация результатов
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# График точности классификации
axes[0, 0].plot(history_class.history['accuracy'], label='Обучающая')
axes[0, 0].plot(history_class.history['val_accuracy'], label='Валидационная')
axes[0, 0].set_title('Точность классификации')
axes[0, 0].set_xlabel('Эпоха')
axes[0, 0].set_ylabel('Точность')
axes[0, 0].legend()
axes[0, 0].grid(True)

# Матрица ошибок классификации
cm = confusion_matrix(y_test_class, y_pred_class_labels)
im = axes[0, 1].imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
axes[0, 1].set_title('Матрица ошибок классификации')
plt.colorbar(im, ax=axes[0, 1])
axes[0, 1].set_xticks([0, 1, 2])
axes[0, 1].set_yticks([0, 1, 2])
axes[0, 1].set_xticklabels(['Бюджет', 'Средний', 'Премиум'])
axes[0, 1].set_yticklabels(['Бюджет', 'Средний', 'Премиум'])

# График потерь регрессии
axes[1, 0].plot(history_reg.history['loss'], label='Обучающая')
axes[1, 0].plot(history_reg.history['val_loss'], label='Валидационная')
axes[1, 0].set_title('Потери регрессии (MSE)')
axes[1, 0].set_xlabel('Эпоха')
axes[1, 0].set_ylabel('MSE')
axes[1, 0].legend()
axes[1, 0].grid(True)

# График предсказанных vs реальных цен
axes[1, 1].scatter(y_test_reg, y_pred_reg, alpha=0.5)
axes[1, 1].plot([y_test_reg.min(), y_test_reg.max()],
                [y_test_reg.min(), y_test_reg.max()], 'r--', lw=2)
axes[1, 1].set_xlabel('Реальная цена (руб.)')
axes[1, 1].set_ylabel('Предсказанная цена (руб.)')
axes[1, 1].set_title('Предсказанные vs реальные цены')
axes[1, 1].grid(True)

plt.tight_layout()
plt.show()

# 7. Пример прогноза для нового автомобиля
print("\n" + "="*50)
print("ПРИМЕР ПРОГНОЗА ДЛЯ НОВОГО АВТОМОБИЛЯ")
print("="*50)

# Характеристики нового автомобиля
new_car = pd.DataFrame({
    'horsepower': [180],
    'mileage': [45000],
    'year': [2020],
    'engine_volume': [2.5],
    'transmission': ['automatic'],
    'fuel_type': ['petrol'],
    'car_type': ['sedan'],
})

# Кодируем категориальные признаки
for column in ['transmission', 'fuel_type', 'car_type']:
    le = label_encoders[column]
    try:
        new_car[column + '_encoded'] = le.transform(new_car[column])
    except ValueError:
        # Если встречается новое значение, используем наиболее частый класс
        new_car[column + '_encoded'] = le.transform([le.classes_[0]])

# Подготовка признаков
new_car_features = new_car[features].values
new_car_scaled = scaler.transform(new_car_features)

# Прогноз классификации
class_prob = model_class.predict(new_car_scaled)
predicted_class = np.argmax(class_prob, axis=1)[0]
class_names = ['Бюджетный', 'Средний класс', 'Премиальный']

# Прогноз регрессии
predicted_price_scaled = model_reg.predict(new_car_scaled)
predicted_price = scaler_price.inverse_transform(predicted_price_scaled)[0][0]

print(f"\nХарактеристики автомобиля:")
print(f"Мощность: {new_car['horsepower'].values[0]} л.с.")
print(f"Пробег: {new_car['mileage'].values[0]:,.0f} км")
print(f"Год выпуска: {new_car['year'].values[0]}")
print(f"Объем двигателя: {new_car['engine_volume'].values[0]} л")
print(f"Коробка передач: {new_car['transmission'].values[0]}")
print(f"Тип топлива: {new_car['fuel_type'].values[0]}")
print(f"Тип кузова: {new_car['car_type'].values[0]}")

print(f"\nПрогноз:")
print(f"Ценовая категория: {class_names[predicted_class]}")
print(f"Вероятности: Бюджет={class_prob[0][0]:.2%}, Средний={class_prob[0][1]:.2%}, Премиум={class_prob[0][2]:.2%}")
print(f"Прогнозируемая цена: {predicted_price:,.0f} руб.")

# 8. Анализ важности признаков (на основе весов первого слоя)
print("\n" + "="*50)
print("АНАЛИЗ ВАЖНОСТИ ПРИЗНАКОВ")
print("="*50)

weights = model_class.layers[0].get_weights()[0]
feature_importance = np.abs(weights).mean(axis=1)

importance_df = pd.DataFrame({
    'Признак': features,
    'Важность': feature_importance
}).sort_values('Важность', ascending=False)

print("\nОтносительная важность признаков для классификации:")
for idx, row in importance_df.iterrows():
    print(f"{row['Признак']}: {row['Важность']:.4f}")

# Корреляционный анализ
correlation_matrix = df[features + ['price']].corr()
print("\nКорреляция признаков с ценой:")
print(correlation_matrix['price'].sort_values(ascending=False))
```
